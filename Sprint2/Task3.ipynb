Описание задачи:
Создать базовый класс SavingsAccount в C#, который будет представлять
специализированные учетные записи для сбережений в банке. На основе этого
класса разработать 2-3 производных класса, демонстрирующих принципы
наследования и полиморфизма. В каждом из классов должны быть реализованы
новые атрибуты и методы, а также переопределены некоторые методы базового
класса для демонстрации полиморфизма.

Требования к базовому классу SavingsAccount:

• Атрибуты: ID счета (AccountId), Баланс (Balance), Процентная ставка
(InterestRate).

• Методы:
o Deposit(): метод для внесения денег на счет.
o Withdraw(): метод для снятия денег со счета.
o CalculateInterest(): метод для расчета процентов по счету.

Требования к производным классам:

1. СтуденческаяУчетнаяЗапись (StudentAccount): Должна содержать
дополнительные атрибуты, такие как Год обучения (YearOfStudy).
Метод CalculateInterest() должен быть переопределен для применения
сниженной процентной ставки для студентов.
2. ПремиумУчетнаяЗапись (PremiumAccount): Должна содержать
дополнительные атрибуты, такие как Минимальный баланс
(MinimumBalance). Метод Withdraw() должен быть переопределен для
ограничения снятия средств до минимального баланса.
3. ИнвестиционныйУчет (InvestmentAccount) (если требуется третий класс):
Должна содержать дополнительные атрибуты, такие как Инвестиционный
портфель (PortfolioValue). Метод Deposit() должен быть переопределен для
автоматического инвестирования части внесенных средств.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

```
using System;
using System.Collections.Generic;

public interface INotifiable
{
    void SendNotification(string message);
}

public interface ITaxable
{
    double CalculateTax();
}

public class SavingsAccount : INotifiable
{
    public string AccountID { get; set; }
    public double Balance { get; set; }
    public double InterestRate { get; set; }
    public string OwnerName { get; set; }
    public DateTime CreationDate { get; set; }

    public SavingsAccount(string _accountID, double _balance, double _interestRate, string _ownerName)
    {
        AccountID = _accountID;
        Balance = _balance;
        InterestRate = _interestRate;
        OwnerName = _ownerName;
        CreationDate = DateTime.Now;
    }

    public virtual void Deposit(double amount)
    {
        if (amount >= 0)
        {
            Balance += amount;
            Console.WriteLine($"[Базовый] Баланс счёта: {Balance}");
        }
        else
        {
            Console.WriteLine("[Базовый] Ошибка пополнения.");
        }
    }

    public virtual void Withdraw(double amount)
    {
        if (amount >= 0 && amount <= Balance)
        {
            Balance -= amount;
            Console.WriteLine($"[Базовый] Остаток: {Balance}");
        }
        else
        {
            Console.WriteLine("[Базовый] Ошибка снятия средств.");
        }
    }

    public virtual double CalculateInterest()
    {
        double interest = Balance * InterestRate;
        Console.WriteLine($"[Базовый] Проценты по счёту: {interest}");
        return interest;
    }

    public virtual void Transfer(SavingsAccount targetAccount, double amount)
    {
        if (amount > 0 && this.Balance >= amount)
        {
            this.Balance -= amount;
            targetAccount.Deposit(amount);
            Console.WriteLine($"[Базовый] Перевод {amount} выполнен.");
        }
        else
        {
            Console.WriteLine("[Базовый] Недостаточно средств для перевода.");
        }
    }

    public int GetAccountAgeInDays()
    {
        return (DateTime.Now - CreationDate).Days;
    }

    public virtual void ApplyMonthlyFee()
    {
        double fee = 100;
        Balance -= fee;
        Console.WriteLine($"[Базовый] Списана комиссия за обслуживание: {fee}. Баланс: {Balance}");
    }

    public virtual void SendNotification(string message)
    {
        Console.WriteLine($"[Уведомление для {OwnerName}]: {message}");
    }
}

public class StudentAccount : SavingsAccount
{
    public string UniversityName { get; set; }

    public StudentAccount(string _accountID, double _balance, double _interestRate, string _ownerName, string _university) 
        : base(_accountID, _balance, _interestRate, _ownerName)
    {
        UniversityName = _university;
    }

    public override double CalculateInterest()
    {
        double StudentInterestRate = InterestRate * 0.15; 
        double StudentInterest = Balance * StudentInterestRate;
        Console.WriteLine($"[Студенческий] Проценты по счёту: {StudentInterest}");
        return StudentInterest; 
    }

    public override void ApplyMonthlyFee()
    {
        Console.WriteLine($"[Студенческий] Комиссия за обслуживание не взимается для студентов {UniversityName}.");
    }
}

public class PremiumAccount : SavingsAccount, ITaxable, INotifiable
{
    public string PersonalManagerName { get; set; }

    public PremiumAccount(string _accountID, double _balance, double _interestRate, string _ownerName, string _managerName) 
        : base(_accountID, _balance, _interestRate, _ownerName)
    {
        PersonalManagerName = _managerName;
    }

    public override void Withdraw(double amount)
    {
        double minimumBalance = 2000;
        if (amount >= 0 && ((Balance - amount) >= minimumBalance))
        {
            Balance -= amount;
            Console.WriteLine($"[Премиум] Остаток: {Balance}");
        }
        else
        {
            Console.WriteLine($"[Премиум] Ошибка: баланс не может быть ниже {minimumBalance}.");
        }
    }

    public double CalculateTax()
    {
        double tax = Balance * 0.01;
        Console.WriteLine($"[Премиум] Рассчитан налог на богатство: {tax}");
        return tax;
    }

    public override void SendNotification(string message)
    {
        Console.WriteLine($"[ПРЕМИУМ УВЕДОМЛЕНИЕ] Уважаемый {OwnerName}, ваш персональный менеджер {PersonalManagerName} сообщает: {message}");
    }
}

public class InvestmentAccount : SavingsAccount
{
    public double PortfolioValue { get; set; }
    public string RiskProfile { get; set; }
    private double InvestmentRatio = 0.25;

    public InvestmentAccount(string _accountID, double _balance, double _interestRate, string _ownerName, double _portfolio, string _riskProfile) 
        : base(_accountID, _balance, _interestRate, _ownerName)
    {
        PortfolioValue = _portfolio;
        RiskProfile = _riskProfile;
    }

    public override void Deposit(double amount)
    {
        if (amount > 0)
        {
            double InvestmentPart = amount * InvestmentRatio;
            double BalancePart = amount - InvestmentPart;
            PortfolioValue += InvestmentPart;
            Balance += BalancePart;
            Console.WriteLine($"[Инвестиционный] Баланс счёта: {Balance}. Баланс портфеля: {PortfolioValue} (Риск: {RiskProfile})");
        }
    }

    public void RebalancePortfolio()
    {
        Console.WriteLine($"[Инвестиционный] Портфель клиента {OwnerName} перебалансирован согласно профилю риска: {RiskProfile}.");
    }
}


 SavingsAccount[] bankClients = new SavingsAccount[]
{
    new StudentAccount("STU-101", 1000, 0.10, "Алексей", "МГУ"),
    new PremiumAccount("PRM-202", 5000, 0.08, "Елена", "Дмитрий"),
    new InvestmentAccount("INV-303", 2000, 0.05, "Олег", 500, "Агрессивный")
};

Console.WriteLine("══════════ 1. Стандартная обработка (Полиморфизм) ══════════");
foreach (var account in bankClients)
{
    Console.WriteLine($"\nОбработка счета ID: {account.AccountID} (Владелец: {account.OwnerName})");
    account.Deposit(1000);
    account.Withdraw(500);
    account.CalculateInterest();
    account.ApplyMonthlyFee();
}

Console.WriteLine("\n══════════ 2. Демонстрация множественного наследования (Интерфейсы) ══════════");
PremiumAccount premiumClient = (PremiumAccount)bankClients[1];
premiumClient.CalculateTax();
premiumClient.SendNotification("Вам одобрен новый кредитный лимит!");

Console.WriteLine("\n══════════ 3. Демонстрация новых методов производных классов ══════════");
InvestmentAccount invClient = (InvestmentAccount)bankClients[2];
invClient.RebalancePortfolio();

Console.WriteLine("\n══════════ 4. Взаимодействие объектов ══════════");
 bankClients[0].Transfer(bankClients[1], 300);

Console.WriteLine("\n═══════════════════════════════════════════════════════════");
```
