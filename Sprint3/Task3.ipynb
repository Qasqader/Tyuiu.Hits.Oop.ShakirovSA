Описание задачи:
Создать базовый класс SavingsAccount в C#, который будет представлять
специализированные учетные записи для сбережений в банке. На основе этого
класса разработать 2-3 производных класса, демонстрирующих принципы
наследования и полиморфизма. В каждом из классов должны быть реализованы
новые атрибуты и методы, а также переопределены некоторые методы базового
класса для демонстрации полиморфизма.

Требования к базовому классу SavingsAccount:

• Атрибуты: ID счета (AccountId), Баланс (Balance), Процентная ставка
(InterestRate).

• Методы:
o Deposit(): метод для внесения денег на счет.
o Withdraw(): метод для снятия денег со счета.
o CalculateInterest(): метод для расчета процентов по счету.

Требования к производным классам:

1. СтуденческаяУчетнаяЗапись (StudentAccount): Должна содержать
дополнительные атрибуты, такие как Год обучения (YearOfStudy).
Метод CalculateInterest() должен быть переопределен для применения
сниженной процентной ставки для студентов.
2. ПремиумУчетнаяЗапись (PremiumAccount): Должна содержать
дополнительные атрибуты, такие как Минимальный баланс
(MinimumBalance). Метод Withdraw() должен быть переопределен для
ограничения снятия средств до минимального баланса.
3. ИнвестиционныйУчет (InvestmentAccount) (если требуется третий класс):
Должна содержать дополнительные атрибуты, такие как Инвестиционный
портфель (PortfolioValue). Метод Deposit() должен быть переопределен для
автоматического инвестирования части внесенных средств.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и прегегрузкой методов, а также generic классы

```
using System;
using System.Collections.Generic;

public interface INotifiable
{
    void SendNotification(string message);
}

public interface ITaxable
{
    double CalculateTax();
}

public class SavingsAccount : INotifiable
{
    public string AccountID { get; set; }
    public double Balance { get; set; }
    public double InterestRate { get; set; }
    public string OwnerName { get; set; }
    public DateTime CreationDate { get; set; }
    public string Currency { get; set; }
    public string PinCode { get; set; }

    public SavingsAccount(string _accountID, double _balance, double _interestRate, string _ownerName, string _currency = "RUB", string _pinCode = "1234")
    {
        AccountID = _accountID;
        Balance = _balance;
        InterestRate = _interestRate;
        OwnerName = _ownerName;
        CreationDate = DateTime.Now;
        Currency = _currency;
        PinCode = _pinCode;
    }

    public virtual void Deposit(double amount)
    {
        if (amount >= 0)
        {
            Balance += amount;
            Console.WriteLine($"[Базовый] Баланс счёта: {Balance} {Currency}");
        }
        else
        {
            Console.WriteLine("[Базовый] Ошибка пополнения.");
        }
    }

    public virtual void Deposit(double amount, string description)
    {
        Deposit(amount);
        Console.WriteLine($"[Базовый] Описание: {description}");
    }

    public virtual void Withdraw(double amount)
    {
        if (amount >= 0 && amount <= Balance)
        {
            Balance -= amount;
            Console.WriteLine($"[Базовый] Остаток: {Balance} {Currency}");
        }
        else
        {
            Console.WriteLine("[Базовый] Ошибка снятия средств.");
        }
    }

    public virtual void Withdraw(string pin, double amount)
    {
        if (pin == PinCode)
        {
            Console.WriteLine($"[Базовый] PIN-код подтвержден.");
            Withdraw(amount);
        }
        else
        {
            Console.WriteLine("[Базовый] Ошибка: неверный PIN-код.");
        }
    }

    public virtual void Withdraw(int amount)
    {
        Console.WriteLine($"[Базовый] Снятие целой суммы: {amount}");
        Withdraw((double)amount);
    }

    public virtual void Withdraw(double amount, bool applyFee)
    {
        Withdraw(amount);
        if (applyFee)
        {
            double fee = 50;
            Balance -= fee;
            Console.WriteLine($"[Базовый] Применена комиссия: {fee} {Currency}. Итоговый баланс: {Balance} {Currency}");
        }
    }

    public virtual double CalculateInterest()
    {
        double interest = Balance * InterestRate;
        Console.WriteLine($"[Базовый] Проценты: {interest} {Currency}");
        return interest;
    }

    public virtual void Transfer(SavingsAccount targetAccount, double amount)
    {
        if (amount > 0 && this.Balance >= amount)
        {
            this.Balance -= amount;
            targetAccount.Deposit(amount);
            Console.WriteLine($"[Базовый] Перевод {amount} {Currency} выполнен.");
        }
        else
        {
            Console.WriteLine("[Базовый] Недостаточно средств.");
        }
    }

    public virtual void ApplyMonthlyFee()
    {
        double fee = 100;
        Balance -= fee;
        Console.WriteLine($"[Базовый] Комиссия: {fee} {Currency}. Баланс: {Balance} {Currency}");
    }

    public virtual void SendNotification(string message)
    {
        Console.WriteLine($"[Уведомление для {OwnerName}]: {message}");
    }
}

public class StudentAccount : SavingsAccount
{
    public string UniversityName { get; set; }
    public int YearOfStudy { get; set; }

    public StudentAccount(string _accountID, double _balance, double _interestRate, string _ownerName, string _university, int _yearOfStudy) 
        : base(_accountID, _balance, _interestRate, _ownerName)
    {
        UniversityName = _university;
        YearOfStudy = _yearOfStudy;
    }

    public override double CalculateInterest()
    {
        double StudentInterestRate = InterestRate * 0.15; 
        double StudentInterest = Balance * StudentInterestRate;
        Console.WriteLine($"[Студенческий] Проценты: {StudentInterest} {Currency}");
        return StudentInterest; 
    }

    public override void ApplyMonthlyFee()
    {
        Console.WriteLine($"[Студенческий] Комиссия не взимается для студентов {UniversityName}.");
    }

    public override void Withdraw(string pin, double amount)
    {
        Console.WriteLine($"[Студенческий] Проверка студенческого билета ({UniversityName}, {YearOfStudy} курс)...");
        base.Withdraw(pin, amount);
    }

    public void GetStudentInfo()
    {
        Console.WriteLine($"[Студенческий] {OwnerName}, {YearOfStudy} курс, {UniversityName}");
    }
}

public class PremiumAccount : SavingsAccount, ITaxable
{
    public string PersonalManagerName { get; set; }
    public double MinimumBalance { get; set; }

    public PremiumAccount(string _accountID, double _balance, double _interestRate, string _ownerName, string _managerName, double _minimumBalance = 2000) 
        : base(_accountID, _balance, _interestRate, _ownerName)
    {
        PersonalManagerName = _managerName;
        MinimumBalance = _minimumBalance;
    }

    public override void Withdraw(double amount, bool applyFee)
    {
        Console.WriteLine($"[Премиум] Снятие без комиссии для премиум-клиентов.");
        Withdraw(amount);
    }

    public override void Withdraw(double amount)
    {
        if (amount >= 0 && ((Balance - amount) >= MinimumBalance))
        {
            Balance -= amount;
            Console.WriteLine($"[Премиум] Остаток: {Balance} {Currency}");
        }
        else
        {
            Console.WriteLine($"[Премиум] Ошибка: баланс не может быть ниже {MinimumBalance} {Currency}.");
        }
    }

    public double CalculateTax()
    {
        double tax = Balance * 0.01;
        Console.WriteLine($"[Премиум] Налог: {tax} {Currency}");
        return tax;
    }

    public override void SendNotification(string message)
    {
        Console.WriteLine($"[ПРЕМИУМ] {OwnerName}, менеджер {PersonalManagerName}: {message}");
    }
}

public class InvestmentAccount : SavingsAccount
{
    public double PortfolioValue { get; set; }
    public string RiskProfile { get; set; }
    private double InvestmentRatio = 0.25;

    public InvestmentAccount(string _accountID, double _balance, double _interestRate, string _ownerName, double _portfolio, string _riskProfile) 
        : base(_accountID, _balance, _interestRate, _ownerName)
    {
        PortfolioValue = _portfolio;
        RiskProfile = _riskProfile;
    }

    public override void Deposit(double amount)
    {
        if (amount > 0)
        {
            double InvestmentPart = amount * InvestmentRatio;
            double BalancePart = amount - InvestmentPart;
            PortfolioValue += InvestmentPart;
            Balance += BalancePart;
            Console.WriteLine($"[Инвестиционный] Баланс: {Balance} {Currency}, Портфель: {PortfolioValue} {Currency}");
        }
    }

    public void Deposit(double amount, string investmentType)
    {
        Deposit(amount);
        Console.WriteLine($"[Инвестиционный] Тип инвестиции: {investmentType}");
    }

    public void RebalancePortfolio()
    {
        Console.WriteLine($"[Инвестиционный] Портфель {OwnerName} перебалансирован (риск: {RiskProfile})");
    }
}

public class BankManager<T> where T : SavingsAccount
{
    private List<T> _accounts;
    private string _bankName;

    public BankManager(string bankName)
    {
        _accounts = new List<T>();
        _bankName = bankName;
    }

    public void AddAccount(T account)
    {
        _accounts.Add(account);
        Console.WriteLine($"[{_bankName}] Добавлен счёт: {account.AccountID}");
    }

    public void DisplayAllAccounts()
    {
        Console.WriteLine($"\n[{_bankName}] Счетов: {_accounts.Count}");
        foreach (var account in _accounts)
        {
            Console.WriteLine($"  - {account.AccountID}: {account.OwnerName}, {account.Balance} {account.Currency}");
        }
    }
}

StudentAccount studentAcc = new StudentAccount("STU-101", 1000, 0.10, "Алексей", "МГУ", 2);
PremiumAccount premiumAcc = new PremiumAccount("PRM-202", 5000, 0.08, "Елена", "Дмитрий");
InvestmentAccount investAcc = new InvestmentAccount("INV-303", 2000, 0.05, "Олег", 500, "Агрессивный");

Console.WriteLine("=== Перегрузка методов Withdraw с разными типами данных ===");
Console.WriteLine("\n--- Withdraw(double) ---");
studentAcc.Withdraw(100.5);

Console.WriteLine("\n--- Withdraw(string pin, double amount) ---");
studentAcc.Withdraw("1234", 200);

Console.WriteLine("\n--- Withdraw(int) ---");
studentAcc.Withdraw(50);

Console.WriteLine("\n--- Withdraw(double, bool) ---");
premiumAcc.Withdraw(500, true);

Console.WriteLine("\n=== Перегрузка методов Deposit ===");
investAcc.Deposit(1000);
investAcc.Deposit(500, "Акции");

Console.WriteLine("\n=== Перекрытие методов (полиморфизм) ===");
SavingsAccount[] accounts = { studentAcc, premiumAcc, investAcc };
foreach (var acc in accounts)
{
    acc.ApplyMonthlyFee();
}

Console.WriteLine("\n=== Generic класс ===");
BankManager<SavingsAccount> bankManager = new BankManager<SavingsAccount>("СберБанк");
bankManager.AddAccount(studentAcc);
bankManager.AddAccount(premiumAcc);
bankManager.AddAccount(investAcc);
bankManager.DisplayAllAccounts();
```
